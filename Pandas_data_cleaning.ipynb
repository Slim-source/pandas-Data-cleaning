```python
df=pd.read_csv("openpowerlifting.csv")
# print(df.info())
# print(df.head())
# print(df.isnull().sum())
df_clean=df.dropna()
df_clean=df.dropna(axis=1)
print(df.isnull().sum())
```

```python
# 1. Keep a copy of the original dataset to clean
df_clean = df.copy()

# 2. Drop rows where key bodyweight or total lift metrics are missing
df_clean = df_clean.dropna(subset=['BodyweightKg', 'TotalKg'])

# 3. Fill missing Age values with the median age of the athletes
median_age = df_clean['Age'].median()
df_clean['Age'] = df_clean['Age'].fillna(median_age)

# 4. Fill missing values in Division with 'General'
df_clean['Division'] = df_clean['Division'].fillna('General')

# 5. Fill missing lift-specific columns with 0 (indicates no attempt/null entry)
lift_cols = ['Squat4Kg', 'BestSquatKg', 'Bench4Kg', 'BestBenchKg', 'Deadlift4Kg', 'BestDeadliftKg', 'Wilks']
df_clean[lift_cols] = df_clean[lift_cols].fillna(0)

# 6. Verify the cleaning results
print("Remaining nulls per column in df_clean:")
print(df_clean.isnull().sum())
print(f"\nCleaned dataset shape: {df_clean.shape}")
display(df_clean.head())
```

```python
# 7. Handle remaining nulls in WeightClassKg and Place
df_clean['WeightClassKg'] = df_clean['WeightClassKg'].fillna('Unknown')
df_clean['Place'] = df_clean['Place'].fillna('Unknown')

# Verify that there are absolutely no null values left in the dataset
print("Final check of nulls in df_clean:")
print(df_clean.isnull().sum())
print(f"\nFinal Cleaned Dataset Shape: {df_clean.shape}")
```
